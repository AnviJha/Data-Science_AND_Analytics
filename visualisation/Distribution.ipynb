```python
#Distribution plot :use for quantitative data
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
# seaborn has some its own dataset , we can see the list of dataset in the site 
#here we are going to load tips daatset 
```

```python
df=sns.load_dataset('tips')
df
```

```python
#total_bill  , tip id only quantitative data 
```

```python
df['size'].unique()
```

```python
#there is only 6 unique size , hence it is qualitative data 
```

```python
#distribution plot 
sns.displot(df['total_bill'])
```

```python
plt.subplot(1,2,1)
#we can use histplot instead of displot for plotting the graph
sns.histplot(df['total_bill'],bins=100, kde=True)
#bins = controls how the data is divided into intervals (groups).range is divided default:20
#Divide → Count → Draw bars
#kde = kernel density estimate :smooth curve that estimates the probability distribution of your data.
plt.subplot(1,2,2)
sns.histplot(df['tip'], bins =20 , kde=True)
```

```python
#now what if we have to join this two data 
#we use join plot for this 
```

```python
sns.jointplot(x='total_bill' , y='tip' , data=df , kind='hex')# bydefault kind: scatter
```

```python
sns.pairplot(df , hue='sex',palette='rainbow')
```

```python
sns.rugplot(df['total_bill'])
```
