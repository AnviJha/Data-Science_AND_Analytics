```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#matplotlib basics 
x = np.linspace(0,10,10)
x
y=x**2
y
```

```python
#plot it 
plt.plot(x,y)
```

```python
#label the axis 

plt.xlabel("age")
plt.ylabel("study hours")
plt.title("Discipline")
plt.plot(x,y)
```

```python
#subplot  rows X columns grid , position of plot 
plt.subplot(2,2,1).plot(y,x)
plt.subplot(2,2,2).plot(x,y)
plt.subplot(2,2,3).plot(x,y)
plt.subplot(2,2,4).plot(y,x)
```

```python
#to create entire gird at once 
fig, axes = plt.subplots(2, 2)
```

```python

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot([1, 2, 3, 4])
axes[0, 0].set_title("Line Plot")

axes[0, 1].bar([1, 2, 3], [10, 20, 15])
axes[0, 1].set_title("Bar Plot")

axes[1, 0].scatter([1, 2, 3], [10, 20, 15])
axes[1, 0].set_title("Scatter Plot")

axes[1, 1].plot([4, 3, 2, 1])
axes[1, 1].set_title("Another Plot")

plt.show()
```

```python
#greate a graph for this two 
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [30, 32, 31, 29, 33]
plt.plot(days , temperature)
plt.title("Weather")
plt.xlabel("days")
plt.ylabel("temperature")
```

```python
#figure : A Figure is the entire canvas/window containing your plots
fig = plt.figure()
plt.show()
```

```python
# Axes : An Axes is an individual plotting area inside the Figure.
#use of axes to plot is called as object oriented approach 
```

```python
# Put a line plot in axes[0] and a bar plot in axes[1].
fig, axes = plt.subplots(1, 2)

# Line plot
axes[0].plot([1, 2, 3, 4])
axes[0].set_title("Line Plot")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")

# Bar plot
axes[1].bar([1, 2, 3], [10, 20, 15])
axes[1].set_title("Bar Plot")
axes[1].set_xlabel("x axis")
axes[1].set_ylabel("y axis")

plt.show()
```

```python
# #
# Matplotlib allows the aspect ratio, DPI and figure size to be specified 
#when the Figure object is created. You can use the figsize and dpi keyword arguments.

# figsize is a tuple of the width and height of the figure in inches dpi 
#is the dots-per-inch (pixel per inch).
```

```python
fig = plt.figure(figsize = (20,8) , dpi=100)
axis1 = fig.add_axes([0.1, 0.1 , .4 , .8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3 , 0.4 ,.3 , .3])
axis2.plot(x,y)
```
