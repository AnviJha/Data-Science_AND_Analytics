```python
import numpy as np
arr=np.array([2,4,7,10,15,20,25,205])
arr
```

```python
Q1 = np.percentile(arr,25)
print(Q1)
```

```python
Q2 =np.percentile(arr,50)
print(Q2)
```

```python
Q3 = np.percentile(arr,75)
print(Q3)
```

```python
IQR = Q3-Q1
print(IQR)
```

```python
upper_fence=Q1 + 1.5*IQR
print(upper_fence)
lower_fence=Q3 - 1.5*IQR
print(lower_fence)
```

```python
l=[]
for i in arr:
   if i>=lower_fence and i<=upper_fence:
        l.append(i)
arr2=np.array(l)
arr2
```

```python

#BOXPLOT 

import seaborn as sns
sns.boxplot(x=arr2)
```
