```python
import numpy as np 
import pandas as pd 
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df= pd.DataFrame(data)
```

```python
df.isna()
```

```python
df.isna().any() # any column with no null value ?
```

```python
#use of threshold 
df
```

```python
df.dropna(thresh=3) #vo rows jisme 3 atleast non null values ho 
```

```python
#filling the missign data 
df.fillna(888888) #fill nan with 8888888
```

```python
#fill by column 
val={"A":111, "B":222,"C":333,"D":444} #to actually change the original value just put the inplace 
df.fillna(value=val)
```
