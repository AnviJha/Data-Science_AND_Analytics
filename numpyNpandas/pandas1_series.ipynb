```python
import numpy as np
import pandas as pd
```

```python
#creating Series : a series is one dimensional labelled array capable of holding any data type .
#the axis labels collectively called the index 

labels =['a' ,'b','c']
my_list=[10,20,30]
arr=np.array([10,20,30])
d = {1:10 , 2:20 , 3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```
