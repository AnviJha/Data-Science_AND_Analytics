```python
import numpy as np
```

```python
arr=np.arange(11,21)
arr
```

```python
print(arr[6]) #indexing value on 6th index 
```

```python
arr[3 : 7 : 2] #slice - arr[start , end+1,step] ,start default:0, end default :end 
```

```python
arr[4:]
```

```python
arr[:5]
```

```python
#numpy indexing and slicing of matrix 
arr= np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0] # fetch row arr[index_row]
```

```python
print(arr[0,0]) #row and col 
```

```python
#in the above array slice out the portion 19 , 20 , 24, 25, 29, 30
print(arr[3: ,3: ]) # arr[initial row : end row , initial col :end col]
```
