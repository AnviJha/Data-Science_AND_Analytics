```python
def plotting(var , num):
    plt.subplot(2,2,num)
    sns.histplot(df[var] , kde=True)
```

```python
plotting('Age',1)
plotting('RestingBP',2)
plotting('Cholesterol',3)
plotting('MaxHR',4)
```

```python
df['Cholesterol'].value_counts()
```




#total of 918 rows  out of which 172 people have zero cholesterol , in real life zero cholesetrol not possible.
#0 generally indicates missing or unavailable cholesterol measurement

```python
#fill out the zero cholesterol with mean value 
```

```python
ch_mean = df.loc[df['Cholesterol'] !=0 , 'Cholesterol'].mean()
df['Cholesterol'] = df['Cholesterol'].replace(0 , ch_mean)
df['Cholesterol'] = df['Cholesterol'] .round(2)
df['Cholesterol'].value_counts()
```

```python
#same for resting BP
resting_mean = df.loc[df['RestingBP'] !=0 , 'RestingBP'].mean()
df['RestingBP'] = df['RestingBP'].replace(0 , resting_mean )
df['RestingBP'] = df['RestingBP'] .round(2)
df['RestingBP'].value_counts()
```
