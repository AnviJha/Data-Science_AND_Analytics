GRADIENT DESCENT 

```python
import numpy as np
import matplotlib.pyplot as plt
```

Training dataset

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])
```

initial parameter 

```python
theta_0 = 0       # intercept
theta_1 = 0       # slope

learning_rate = 0.01
epochs = 1000

m = len(x)
```

```python
for i in range(epochs):

    # Prediction
    y_pred = theta_0 + theta_1 * x

    # Error
    error = y_pred - y

    # Gradients
    d_theta_0 = (1 / m) * np.sum(error)
    d_theta_1 = (1 / m) * np.sum(error * x)

    # Update parameters
    theta_0 = theta_0 - learning_rate * d_theta_0
    theta_1 = theta_1 - learning_rate * d_theta_1
```

```python
# -------------------------
# 4. Final parameters
# -------------------------
print("Theta 0:", theta_0)
print("Theta 1:", theta_1)

print("Equation:")
print(f"y = {theta_0:.2f} + {theta_1:.2f}x")
```

```python
# -------------------------
# 5. Plot
# -------------------------
y_pred = theta_0 + theta_1 * x

plt.scatter(x, y)
plt.plot(x, y_pred)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression using Gradient Descent")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

# Initial parameters
theta_0 = 0
theta_1 = 0

learning_rate = 0.01
epochs = 100

m = len(x)

# Store values for plotting
cost_history = []
theta1_history = []

for i in range(epochs):

    # Prediction
    y_pred = theta_0 + theta_1 * x

    # Error
    error = y_pred - y

    # Cost
    cost = (1 / (2 * m)) * np.sum(error ** 2)

    # Store
    cost_history.append(cost)
    theta1_history.append(theta_1)

    # Gradients
    d_theta_0 = (1 / m) * np.sum(error)
    d_theta_1 = (1 / m) * np.sum(error * x)

    # Update
    theta_0 = theta_0 - learning_rate * d_theta_0
    theta_1 = theta_1 - learning_rate * d_theta_1


# Plot Cost vs Iterations
plt.plot(range(epochs), cost_history)

plt.xlabel("Iterations")
plt.ylabel("Cost J(θ)")
plt.title("Gradient Descent")

plt.show()
```
